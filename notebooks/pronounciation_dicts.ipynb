```python
from dotenv import load_dotenv
from elevenlabs.client import ElevenLabs
from elevenlabs import play, PronunciationDictionaryVersionLocator
import os
import requests 


load_dotenv()
elevenlabs = ElevenLabs(
    api_key=os.getenv("ELEVENLABS_API_KEY")
)

# add rules from file and get the uploaded result

with open("dictionary.pls", "r") as f:
    pronunciation_dict = elevenlabs.pronunciation_dictionaries.create_from_file(
        file=f.read(), name="example"
    )

audio_1 = elevenlabs.text_to_speech.convert(
    text="Tomato",
    voice_id="sDuUJMeNJR828mXTRrDh",
    model_id="eleven_turbo_v2",
)

audio_2 = elevenlabs.text_to_speech.convert(
    text="With the dictionary: tomato",
    voice_id="sDuUJMeNJR828mXTRrDh",
    model_id="eleven_turbo_v2",
    pronunciation_dictionary_locators=[
        PronunciationDictionaryVersionLocator(
            pronunciation_dictionary_id=pronunciation_dict.id,
            version_id=pronunciation_dict.version_id,
        )
    ],
)

play.play(audio_1)
play.play(audio_2)
```

Remove rules from the dict

```python
# remove rules first (you already did this)
removed = elevenlabs.pronunciation_dictionaries.rules.remove(
    pronunciation_dictionary_id=pronunciation_dict.id,
    rule_strings=["tomato", "Tomato"],
)

# ✅ add a phoneme rule using plain dicts
dict_added = elevenlabs.pronunciation_dictionaries.rules.add(
    pronunciation_dictionary_id=removed.id,   # same dict id
    rules=[
        {
            "type": "phoneme",               # <- important
            "string_to_match": "tomato",
            "phoneme": "t ə ˈ m eɪ t oʊ",   # IPA example; adjust to taste
            "alphabet": "ipa",
            # optional but useful:
            "match_method": "exact",         # "exact" | "word_boundary" | "regex" (varies by SDK)
            "case_sensitive": False,
            "priority": 0,
        }
    ],
)

# then synth using the NEW version returned by `rules.add`
audio_4 = elevenlabs.text_to_speech.convert(
    text="With the custom phoneme: tomato",
    voice_id="sDuUJMeNJR828mXTRrDh",
    model_id="eleven_turbo_v2",
    pronunciation_dictionary_locators=[
        PronunciationDictionaryVersionLocator(
            pronunciation_dictionary_id=dict_added.id,
            version_id=dict_added.version_id,   # <- new version after adding the rule
        )
    ],
)

play.play(audio_4)
```
